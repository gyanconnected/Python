```python
int("10")       # 10        text -> int
float("2.5")    # 2.5       text -> float
str(5)          # "5"       number -> text
bool(0)         # False     0 is False, other numbers are True
list("abc")     # ['a', 'b', 'c']
set([1, 1, 2])  # {1, 2}    removes duplicates
tuple([1, 2])   # (1, 2)
```
