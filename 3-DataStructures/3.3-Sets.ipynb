```python
# 11.1   Creating Sets​

my_set = {1, 2, 3, 4, 5}        # curly brackets
empty = set()                    # empty set
from_list = set([1, 2, 3, 6, 5, 4, 5, 6])
print(from_list)                 # {1, 2, 3, 4, 5, 6}  duplicates removed
```

```python
# 11.2   Adding and removing​

my_set = {1, 2, 3, 4, 5, 6}

my_set.add(7)       # {1, 2, 3, 4, 5, 6, 7}
my_set.add(7)       # no change, 7 is already there (no error)

my_set.remove(3)    # removes 3
# my_set.remove(10)   # KeyError, because 10 is not in the set
my_set.discard(11)  # no error even though 11 is not in the set

my_set.pop()        # removes and returns ONE arbitrary item
my_set.clear()      # empties the set -> set()
```

```python
# 11.3   Membership test​

my_set = {1, 2, 3, 4, 5}
print(3 in my_set)    # True
print(10 in my_set)   # False
```

```python
# 11.4   Mathematical operations​

# Use these two sets throughout:

set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}

# i) Union (everything from both, no repeats):​

set1.union(set2)         # {1, 2, 3, 4, 5, 6, 7, 8, 9}

# ii) Intersection (only what is in both):​

set1.intersection(set2)  # {4, 5, 6}

# iii) Difference (in the first set but not in the second):​

set1.difference(set2)    # {1, 2, 3}
set2.difference(set1)    # {7, 8, 9}

# iv) Symmetric difference (in one set or the other, but not in both):​

set1.symmetric_difference(set2)   # {1, 2, 3, 7, 8, 9}

# v) intersection_update does the same as intersection, but changes set1 itself instead of creating a new set:​

set1.intersection_update(set2)
print(set1)    # {4, 5, 6}   set1 is now changed
```

```python
# 11.5   Subset and superset​

set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5}

set1.issubset(set2)     # False -> is set1 completely inside set2? No
set1.issuperset(set2)   # True  -> does set1 contain all of set2? Yes
```

```python
# 11.6   Practical uses​

# Example: Remove duplicates from a list:​

lst = [1, 2, 2, 3, 4, 4, 5]
set(lst)      # {1, 2, 3, 4, 5}
```
