```python
# 12.1   Creating dictionaries​

empty_dict = {}        # empty dictionary
empty_dict = dict()    # same thing
student = {"name": "Krish", "age": 32, "grade": 24}

#Keys must be unique. If you repeat a key, the last value wins:

student = {"name": "Krish", "age": 32, "name": 24}
print(student)   # {'name': 24, 'age': 32}

# The second "name" replaced the first one.
```

```python
# 12.2   Accessing elements​

student = {"name": "Krish", "age": 32, "grade": "A"}

print(student["grade"])        # 'A'
print(student["age"])          # 32


# get() is the safer way:​

print(student.get("grade"))                      # 'A'
print(student.get("last_name"))                  # None (no error)
print(student.get("last_name", "Not Available")) # 'Not Available' (custom default)


```

```python
# 12.3   Modifying elements​

# Dictionaries are mutable, so you can update, add, and delete.​
student = {"name": "Krish", "age": 32, "grade": "A"}
student["age"] = 33            # update existing key
student["address"] = "India"   # add a new key
del student["grade"]           # delete a key-value pair

print(student)   

# The same syntax student[key] = value updates if the key exists and adds if it doesn’t.

```

```python
# 12.4   Dictionary methods​

student.keys()     # dict_keys(['name', 'age', 'address'])
student.values()   # dict_values(['Krish', 33, 'India'])
student.items()    # dict_items([('name', 'Krish'), ('age', 33), ('address', 'India')])

```

```python
# 12.5   Copying (important)​
# Wrong way: this is not a copy:​

student = {"name": "Krish", "age": 32, "grade": "A"}
student_copy = student
student["name"] = "Krish2"
print(student_copy)   # also shows 'Krish2'!

# Right way: .copy():​

student_copy1 = student.copy()
student["name"] = "KRish3"
print(student_copy1)  # still 'Krish2'
print(student)        # 'KRish3'
```

```python
# 12.6   Iterating  (looping)​

student = {"name": "Krish", "age": 32, "grade": "A"}

for key in student.keys():
    print(key)              # name, age, address

for value in student.values():
    print(value)            # KRish3, 33, India

for key, value in student.items():
    print(f"{key}:{value}") # name:KRish3, age:33, address:India

# Use .items() when you need both the key and the value.​
```

```python
# 12.7   Nested dictionaries  (A dictionary inside a dictionary)​

students = {
    "student1": {"name": "Krish", "age": 32},
    "student2": {"name": "Gyan", "age": 30}
}

print(students["student2"]["name"])   # 'Peter'
print(students["student2"]["age"])    # 35

# Use two keys: the first picks the inner dictionary, the second picks the value inside it.​
```

```python
# 12.8   Dictionary comprehension​

squares = {x: x**2 for x in range(5)}
print(squares)

# Read it as: “for each x from 0 to 4, make key x with value x squared.”
```
