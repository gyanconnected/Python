```python
# Creating tuples

empty_tuple = ()          # empty tuple
tpl = tuple()             # same thing, using the constructor
numbers = tuple([1,2,3])  # convert a list into a tuple
mixed = (1, "Hello World", 3.14, True)
```

```python
# Accessing elements​

numbers = (1, 2, 3, 4, 5, 6)
numbers[2]      # 3  (index starts at 0)
numbers[-1]     # 6  (last item)
numbers[0:4]    # (1, 2, 3, 4)  (start included, end excluded) (Start, Stop)

numbers[::1]    # (1, 2, 3, 4, 5, 6)  forward
numbers[::-1]   # (6, 5, 4, 3, 2, 1)  backward
numbers[::2]    # (1, 3, 5)           forward, skipping one
```

```python
# Tuples operations

# Concatenation (+): joining two tuples

a = (1, 2, 3)
b = (4, 5, 6)

c = a + b
print(c)

# It works with different types too:

names = ("Amit", "Riya")
ages = (25, 30)

print(names + ages)

# You can only join tuple + tuple:
# (1, 2, 3) + 4        # TypeError
(1, 2, 3) + (4,)     # (1, 2, 3, 4)  works

# Repetition (*): repeating a tuple

x = (1, 2, 3)

print(x * 2)
print(x * 3)

# Another example:

zeros = (0,) * 4
print(zeros)
```

```python
# 10.4   Immutability (the key idea)​

lst = [1,2,3,4,5]
lst[1] = "Krish"      # works, lists are mutable

# numbers[1] = "Krish"  # TypeError: 'tuple' object does not support item assignment

# Real-life example: coordinates

places = {
    (28.61, 77.20): "Delhi",
    (19.07, 72.87): "Mumbai",
    (25.59, 85.13): "Patna",
}

print(places[(25.59, 85.13)])

# Example 1: List ko key banane ki koshish

# d = {[1, 2]: "A"} # TypeError: unhashable type: 'list'

# Error aata hai kyunki list badal sakti hai.
```

```python
# Tuples methods

# Because tuples can’t change, they have only two methods:

numbers.count(1)   # 1 -> how many times the value 1 appears
numbers.index(3)   # 2 -> the index where 3 is first found
```

```python
# Packing: putting values together into a tuple (brackets optional):

packed_tuple = 1, "Hello", 3.14   
print(packed_tuple)
```
