```python
# 13.1   Defining and calling a function​

# def function_name(parameters):
#     """Docstring"""
#     # function body
#     return expression

# Why functions? The video’s example:

# Without a function, you rewrite the even/odd check every time:

num = 24
if num % 2 == 0:
    print("the number is even")
else:
    print("the number is odd")

# With a function, you write it once:

def even_or_odd(num):
    """This function finds even or odd"""
    if num % 2 == 0:
        print("the number is even")
    else:
        print("the number is odd")

even_or_odd(24)   # the number is even
even_or_odd(7)    # the number is odd
```

```python
# 13.2   Multiple parameters and return​

def add(a, b):
    return a + b

result = add(2, 4)
print(result)   # 6

# return sends the answer back so you can store it in a variable (result).​

# ​ print vs return (common confusion)​

def add_print(a, b):
    print(a + b)      # only shows on screen

def add_return(a, b):
    return a + b      # gives the value back

x = add_print(2, 4)   # prints 6, but x is None
y = add_return(2, 4)  # y is 6, you can use it later
```

```python
# 13.3   Default parameters​

def greet(name="Guest"):
    print(f"Hello {name} Welcome To the paradise")

greet("Krish")   # Hello Krish Welcome To the paradise
greet()          # Hello Guest Welcome To the paradise

# A default value is used when the caller doesn’t give one.​
```

```python
# 13.4   Variable-length arguments​

# Sometimes you don’t know how many inputs will come.

# *args: any number of positional arguments​

def print_numbers(*args):
    for number in args:
        print(number)

print_numbers(1, 2, 3, "Krish")

# *args collects all the values into a tuple: (1, 2, 3, "Gyan"). ​

# **kwargs: any number of keyword arguments​

def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}:{value}")

print_details(name="Gyan", age="32", country="India")
```

```python
# 13.5   Returning multiple values​

def multiply(a, b):
    return a * b, a

multiply(2, 3)   # (6, 2)

# Python packs the values into a tuple (6, 2). You can unpack it directly:

product, first = multiply(2, 3)
print(product)   # 6
print(first)     # 2
```
