```python
# 15.  Map Function​

# map()  applies a function to every item of a list (or any iterable) and gives you the results.​

# Stntax

# map(function, iterable)

# Think of it as a machine on a conveyor belt: each item goes in, the function changes it, and the changed item comes out.​

# [1, 2, 3, 4]  -->  function (x*x)  -->  [1, 4, 9, 16]
```

```python
# 15.1   Basic use with normal function​

def square(x):
    return x * x

numbers = [1, 2, 3, 4, 5, 6, 7, 8]
print(list(map(square, numbers)))   # [1, 4, 9, 16, 25, 36, 49, 64]

result = map(square, numbers)
print(result)         # <map object at 0x...>
print(list(result))   # [1, 4, 9, ...]
```

```python
# 15.2   With a Lambda​

list(map(lambda x: x * x, numbers)) 
```

```python
# 15.3 Multiple iterables​

# If the function takes two inputs, pass two lists. map() pairs items by position:​

numbers1 = [1, 2, 3]
numbers2 = [4, 5, 6]

list(map(lambda x, y: x + y, numbers1, numbers2))   # [5, 7, 9]
```

```python
# 15.4 Using build-in functions​

# Example: Convert strings to integers:​

str_numbers = ['1', '2', '3', '4', '5']
list(map(int, str_numbers))   # [1, 2, 3, 4, 5]

# Example: Make words uppercase​
words = ['apple', 'banana', 'cherry']
list(map(str.upper, words))   # ['APPLE', 'BANANA', 'CHERRY']
```

```python
# 15.5 Extracting data list of dictionaries​

def get_name(person):
    return person['name']

people = [
    {'name': 'Krish', 'age': 32},
    {'name': 'Jack', 'age': 33}
]

list(map(get_name, people))   # ['Krish', 'Jack']
```

```python
# 15.6 map() vs a list comprehension​

# Both do the same job, but prefer the comprehension for readability:​
list(map(lambda x: x * x, numbers))   # map
[x * x for x in numbers]              # list comprehension
```
