```python
# 14.1   def vs lambda: same thing, two styles​

# normal function
def addition(a, b):
    return a + b

# lambda version
addition = lambda a, b: a + b

print(addition(5, 6))   # 11
```

```python
# 14.2   A small difference in the even example​
# The two versions don’t behave exactly the same for odd numbers:​

# even(7)    # None   (the def version returns nothing for odd numbers)
# even1(7)   # False  (the lambda always gives True or False)
```

```python
# 14.4   Where lambda is really used: map()​

# map(function, list) applies a function to every item of a list.​

# With a normal function:​

numbers = [1, 2, 3, 4, 5, 6]

def square(number):
    return number ** 2

print(list(map(square, numbers)))   # [1, 4, 9, 16, 25, 36]

# With a lambda (no need to define a separate function):

print(list(map(lambda x: x**2, numbers)))   # [1, 4, 9, 16, 25, 36]
```
