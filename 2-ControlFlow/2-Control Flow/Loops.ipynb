```python
for i in range(5):
    print(i)        # prints 0 1 2 3 4
```

```python
for ch in "Gyam Prakash":
    print(ch)       # G, y, a, m, (space), P, r, a, k, a, s, h
```

```python
count = 0

while count < 5:
    print(count)
    count = count + 1
```

```python
for i in range(10):
    if i == 5:
        break
    print(i)        # 0 1 2 3 4
```

```python
for i in range(10):
    if i % 2 == 0:
        continue    # even number, skip the print
    print(i)        # 1 3 5 7 9
```

```python
for i in range(5):
    if i == 3:
        pass
    print(i)        # 0 1 2 3 4 (3 is still printed)
```

```python
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```
